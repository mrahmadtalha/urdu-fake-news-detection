```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Display settings
pd.set_option('display.max_columns', None)
pd.set_option('display.max_colwidth', 100)

print("Libraries loaded successfully")
```

```python
# Load datasets
df_roman = pd.read_csv('../data/raw/rufnd_roman_urdu.csv')
df_amjad = pd.read_excel('../data/raw/amjad_urdu.xlsx')
df_urdu = pd.read_csv('../data/raw/Urdu_News_Data.csv')

print("RUFND Roman Urdu:", df_roman.shape)
print("Amjad Urdu:", df_amjad.shape)
print("Urdu News Data:", df_urdu.shape)
```

```python
# Preview each dataset
print("=== RUFND Roman Urdu ===")
print(df_roman.head(3))
print("\nColumns:", df_roman.columns.tolist())

print("\n=== Amjad Urdu ===")
print(df_amjad.head(3))
print("\nColumns:", df_amjad.columns.tolist())

print("\n=== Urdu News Data ===")
print(df_urdu.head(3))
print("\nColumns:", df_urdu.columns.tolist())
```

```python
# Check label values in each dataset
print("=== RUFND Labels ===")
print(df_roman['Label'].value_counts())

print("\n=== Amjad Labels ===")
print(df_amjad['label'].value_counts())

print("\n=== Urdu News Labels ===")
print(df_urdu['Label'].value_counts())
```

```python
# Check missing values
print("=== RUFND Missing Values ===")
print(df_roman.isnull().sum())

print("\n=== Amjad Missing Values ===")
print(df_amjad.isnull().sum())

print("\n=== Urdu News Missing Values ===")
print(df_urdu.isnull().sum())
```
